#### **Pydantic**

**Pydantic** is a Python library used for **data validation, parsing and serialization** using Python type hints.

It is especially important in **FastAPI**, because FastAPI uses Pydantic models to validate request bodies and structure API responses.

**Example:**

Install the package using this:
```python
!pip install pydantic
```

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int
    email: str

user = User(name="Shourov Roy", age=23, email="shourov@example.com")
print(user)
````

---
Pydantic checks that the supplied data matches the declared types.

#### **What happens with invalid data?**

````python
user = User(name="Shourov Roy", age="twenty three", email="shourov@example.com")
print(user)
````

Pydantic raises a **validation error** because `"hello"` cannot be interpreted as an integer.
