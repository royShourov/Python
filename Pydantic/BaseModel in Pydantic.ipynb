#### **`BaseModel` in Pydantic**

`BaseModel` is the **core class of Pydantic**. You use it to define the **structure, types, validation rules and behavior of your data**.

---

**Example:**

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int
    email: str

user = User(
    name="Shourov Roy",
    age=23,
    email="shourov@example.com"
)
print(user)
````

Here `class User(BaseModel):` means `User` is a Pydantic model that inherits from `BaseModel`.

---

#### **Why Do We Need `BaseModel`?**

Normally, Python allows this:

```python
user = {
    "name": "Shourov",
    "age": "25"
}
```

Notice that `age` is a string `"25"`. But we want `25`. With Pydantic:

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

user = User(
    name="Shourov Roy",
    age=25
)
print(user.age)
print(type(user.age))
````

Pydantic validates and, where appropriate, converts input according to the declared types.

---

#### **Validation**

One of the most important purposes of `BaseModel` is **data validation**.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

# This is valid
user = User(
    name="Shourov Roy",
    age=23
)
print(user)

# This is invalid
'''''
user = User(
    name="Shourov Roy",
    age="twenty three"
)
'''''
````

Pydantic raises a validation error because `"twenty three"` cannot be interpreted as an integer.

---

#### **Required Fields**

By default, fields without a default value are required.

```python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int
```

This is invalid: `user = User(name="Shourov")`. Because `age` is required.

---

#### **Optional Fields**

You can provide a default value:

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int = 18

user = User(name="Shourov Roy")
print(user.age)
````

You can also explicitly allow `None`:

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int | None = None

user = User(name="Shourov Roy")
print(user.age)
````

---

#### **Nested Models**

One `BaseModel` can contain another `BaseModel`.

````python
from pydantic import BaseModel

class Address(BaseModel):
    city: str
    country: str

class User(BaseModel):
    name: str
    age: int
    address: Address

user = User(
    name="Shourov Roy",
    age= 23,
    address={
        "city": "Dinajpur",
        "country": "Bangladesh"
    }
)

print(user.address.city)
````

---

#### **Convert Model to Dictionary**

Use `model_dump()`. `model_dump()` is the modern Pydantic v2 approach. Example:

````python
# using previous model
data = user.model_dump()
print(data)
````

---

#### **Convert Model to JSON** 

Use `model_dump_json()`. Example:


````python
print(user.model_dump_json())
````

---

#### **Create a Model from a Dictionary**

You can pass a dictionary directly:

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

# This is a dictionary
data = {
    "name": "Shourov Roy",
    "age": 23
}

# Converting dictionary into pydantic model
user = User(**data)
print(user)
````

---

#### **Validate External Data**

Pydantic v2 provides `model_validate()`. Example:

````python
data = {
    "name": 'Shourov Roy',
    "age": 23
}

user = User.model_validate(data)
print(user)
````

This is particularly useful when data comes from:

* APIs
* databases
* JSON
* configuration files
* external services

---

#### **Validation Error**

Pydantic gives structured validation errors.

````python
from pydantic import BaseModel, ValidationError

class User(BaseModel):
    name: str
    age: int

try: 
    user = User(
        name='Shourov Roy',
        age="abc"
    )
except ValidationError as e:
    print(e)
````

You'll get information describing which field failed and why.

#### **`BaseModel` in FastAPI**

This is where Pydantic becomes especially important for your FastAPI learning.

For example:

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class User(BaseModel):
    name: str
    age: int
    email: str

@app.post("/users")
def create_user(user: User):
    return user
```
A client can send:

```json
{
    "name": "Shourov",
    "age": 25,
    "email": "shourov@example.com"
}
```

FastAPI uses the Pydantic `User` model to:

1. Read the request body
2. Parse the incoming data
3. Validate the fields
4. Convert compatible values
5. Provide the validated Python object to your function
6. Generate API documentation

#### **`BaseModel` Is Not Just a Class**

Consider:

```python
class User(BaseModel):
    name: str
    age: int
```

It gives you much more than a normal Python class. A Pydantic model provides things such as:

```text
Type information
       ↓
Validation
       ↓
Parsing
       ↓
Serialization
       ↓
Error reporting
       ↓
JSON Schema
       ↓
FastAPI integration
```


---

**Example:** Suppose you're building an ML prediction API.

```python
from pydantic import BaseModel

class PredictionRequest(BaseModel):
    age: int
    salary: float
    experience: int
```

FastAPI endpoint:

```python
@app.post("/predict")
def predict(data: PredictionRequest):

    age = data.age
    salary = data.salary
    experience = data.experience

    # ML model prediction here
    return {
        "age": age,
        "salary": salary,
        "experience": experience
    }
```

The Pydantic model defines the expected input:

```text
PredictionRequest
├── age: int
├── salary: float
└── experience: int
```

This pattern is extremely common in **FastAPI + Machine Learning APIs**.
