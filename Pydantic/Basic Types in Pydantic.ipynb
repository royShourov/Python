#### **Basic Types in Pydantic**

In Pydantic, **types** tell Pydantic what kind of data a field should contain. Pydantic then **validates** the incoming data and, in many cases **converts compatible values** to the declared type.

---
**`str` — String**: Used for text.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    email: str

user = User(name="Shourov Roy", email="shourov@example.com")
print(user)
````

If you provide an incompatible value:

```python
user = User(name=123, email="test@example.com")
```

Depending on the Pydantic version/configuration, Pydantic may reject the value rather than blindly converting it.

---
**`int` — Integer:** Used for whole numbers.

````python
from pydantic import BaseModel

class Product(BaseModel):
    price: int

product = Product(price=100)
print(product.price)
````

---
**`float` — Decimal Numbers:** Used for numbers containing decimal values.

````python
from pydantic import BaseModel

class Product(BaseModel):
    price: float

product = Product(price=99.99)
print(product.price)
````

Pydantic can also accept compatible numeric input depending on its validation rules.

---

**`bool` — Boolean:** Used for `True` or `False`.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    is_active: bool

user = User(name="Shourov Roy", is_active=True)
print(user.is_active)
````

---
**`list`:** Used when a field contains multiple values. For example, a list of strings:

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    skills: list[str]

user = User(
    name="Shourov Roy",
    skills=["Python", "ML", "Dl"]
)
print(user.skills)
````

---
**`dict`:** Used for key-value data.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    profile: dict

user = User(
    name='Shourov Roy',
    profile={
        "country": "BD",
        "language": "Python"
    }
)
print(user.profile)
````

You can make the type more specific:

```python
class User(BaseModel):
    profile: dict[str, str]
```

Now the expected structure is:

```python
{
    "country": "Bangladesh",
    "language": "Python"
}
```

---
**`tuple`:** Used for a fixed or ordered collection.


````python
from pydantic import BaseModel

class Point(BaseModel):
    coordinates: tuple[float, float]

point = Point(coordinates=(23.81, 90.5))
print(point.coordinates)
````

---
**`set`:** Used when you want unique values.

````python
from pydantic import BaseModel

class User(BaseModel):
    skills: set[str]

user = User(skills=["Python", "FastAPI", "Python"])
print(user.skills)
````

The duplicate `"Python"` is removed because a set contains unique values.

---
**`None` / Optional Values:** Sometimes a field can contain a value **or `None`**. Modern Python uses:

```python
str | None
```


````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    phone: str | None

user = User(name="Shourov", phone=None)
print(user)
````

You can also use:

```python
from typing import Optional

class User(BaseModel):
    phone: Optional[str]
```

For modern Python code, `str | None` is generally preferred. There is a difference between **allowing `None`** and **making a field optional/missing**. For example:

```python
phone: str | None
```

allows `None`, but depending on the Pydantic version and declaration, the field may still be required. If you want the field to be optional and default to `None`:

```python
class User(BaseModel):
    name: str
    phone: str | None = None
```

Now this is valid: `user = User(name="Shourov")`

---
**`datetime`:** Pydantic supports Python's `datetime` type.

````python
from datetime import datetime
from pydantic import BaseModel

class Event(BaseModel):
    name: str
    created_at: datetime

event = Event(
    name="Python Workshop",
    created_at="2026-09-21T15:30:00"
)

print(event.created_at)
````

Pydantic can parse compatible date/time strings into a `datetime`.

---
**`date`:** For dates without a time:

````python
from datetime import date
from pydantic import BaseModel

class User(BaseModel):
    name: str
    birth_date: date

user = User(name="Shourov", birth_date="2000-05-10")
print(user.birth_date)
````

---
**`Decimal`:** Useful when you need decimal arithmetic, especially for financial values.

````python
from decimal import Decimal
from pydantic import BaseModel

class Product(BaseModel):
    price: Decimal

product = Product(price="99.99")
print(product.price)
````

---
**`Any`:** `Any` means essentially **any Python value is accepted**.

````python
from typing import Any
from pydantic import BaseModel

class Data(BaseModel):
    value: Any

data1 = Data(value="Hello")
data2 = Data(value=123)
data3 = Data(value=[1, 2, 3])
````

All can be valid. Use `Any` carefully because you lose much of the benefit of type validation.

---
**Multiple Types with `Union`:** You can allow more than one type. Modern Python:

```python
class Product(BaseModel):
    id: int | str
```

Both can be accepted:

```python
Product(id=101)
Product(id="ABC101")
```

Older style:

```python
from typing import Union

class Product(BaseModel):
    id: Union[int, str]
```


---
**Basic Types:**

| Python Type | Example                 | Typical Use                  |
| ----------- | ----------------------- | ---------------------------- |
| `str`       | `"Shourov"`             | Text                         |
| `int`       | `25`                    | Whole numbers                |
| `float`     | `99.99`                 | Decimal numbers              |
| `bool`      | `True`                  | Boolean values               |
| `list`      | `[1, 2, 3]`             | Multiple ordered values      |
| `dict`      | `{"name": "John"}`      | Key-value data               |
| `tuple`     | `(10, 20)`              | Ordered fixed structure      |
| `set`       | `{"Python", "FastAPI"}` | Unique values                |
| `date`      | `2026-09-21`            | Dates                        |
| `datetime`  | `2026-09-21 15:30`      | Date + time                  |
| `Decimal`   | `Decimal("99.99")`      | Precise decimal values       |
| `Any`       | Anything                | No specific type restriction |
| `T \| None` | `"abc"` / `None`        | Nullable fields              |
